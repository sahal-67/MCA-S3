```python
import numpy as np

A = np.array([[4, 11],
              [14, 5]], dtype=float)

# Compute SVD
U, s, Vt = np.linalg.svd(A)

# Create Sigma matrix
Sigma = np.diag(s)

# Reconstruct A
A_reconstructed = U @ Sigma @ Vt

# Print results
print("U:")
print(np.round(U, 4))

print("\nSingular Values:")
print(np.round(s, 4))

print("\nVt:")
print(np.round(Vt, 4))

print("\nReconstructed A:")
print(np.round(A_reconstructed, 4))
```

```python
import matplotlib.pyplot as plt

groups = ["G1", "G2", "G3", "G4", "G5"]
men = [22, 30, 35, 35, 26]
women = [25, 32, 30, 35, 29]

x = [0, 1, 2, 3, 4]

plt.bar(x, men, width=0.4, label="Men", color="red")
plt.bar([i + 0.4 for i in x], women, width=0.4, label="Women", color="orange")

plt.xticks([i + 0.2 for i in x], groups)
plt.xlabel("Groups")
plt.ylabel("Scores")
plt.title("Scores by Group and Gender")
plt.legend()

plt.show()
```
